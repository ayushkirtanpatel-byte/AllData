## Q1. Explain Mean, Median, and Mode with a real-life example.

Mean, Median, and Mode are the three most commonly used measures of central tendency. They help summarize a dataset by identifying its central or typical value.

- Mean is the arithmetic average of all observations.
- Median is the middle value when data is arranged in order.
- Mode is the value that appears most frequently.

For example, consider monthly salaries of ₹20,000, ₹25,000, ₹30,000, ₹30,000, and ₹100,000.

Mean = ₹41,000, Median = ₹30,000, Mode = ₹30,000.

The mean is affected by the extremely high salary, while the median remains stable. Therefore, median is often preferred when outliers are present in the data.

## Q2. What is the difference between Variance and Standard Deviation?

Variance and Standard Deviation are statistical measures used to describe the spread of data around the mean.

Variance measures the average squared deviation from the mean. Since the differences are squared, the result is expressed in squared units, which can be difficult to interpret.

Standard Deviation is the square root of variance. It is expressed in the same unit as the original data, making it easier to understand and interpret.

A larger standard deviation indicates greater variability in data, while a smaller standard deviation indicates that observations are clustered closer to the mean.

## Q3. Define Normal Distribution and give one practical use case.

A Normal Distribution is a continuous probability distribution represented by a symmetric bell-shaped curve. In a perfect normal distribution, the mean, median, and mode are equal.

Most observations are concentrated near the center, while fewer observations occur toward the extremes. The normal distribution follows the 68-95-99.7 rule.

Examples of normally distributed data include human height, IQ scores, and examination scores. In educational analytics, normal distribution helps teachers analyze overall student performance.

## Q4. Explain Skewness and Kurtosis in simple words.

Skewness measures the asymmetry of a distribution.

- Positive Skewness: Long tail toward the right side.
- Negative Skewness: Long tail toward the left side.
- Zero Skewness: Perfectly symmetric distribution.

Kurtosis measures the shape and peakedness of a distribution.

- High Kurtosis indicates more extreme values and sharper peaks.
- Low Kurtosis indicates flatter distributions with fewer outliers.

These measures help analysts understand the shape of data beyond basic averages.

## Q5. What is Probability? Differentiate between Empirical and Theoretical Probability.

Probability is a numerical measure that describes the likelihood of an event occurring. Its value always lies between 0 and 1.

Theoretical Probability is calculated mathematically before performing an experiment.

Formula:
P(E) = Favorable Outcomes / Total Outcomes

Empirical Probability is based on actual observations or experiments.

Formula:
P(E) = Number of Times Event Occurs / Total Trials

For example, the theoretical probability of obtaining heads from a fair coin is 0.5. If a coin is tossed 100 times and heads appears 48 times, the empirical probability becomes 0.48.

## Q6. Explain Independent vs Dependent Events with one example each.

Independent events are events where the occurrence of one event does not affect the occurrence of another event.

Example:
Rolling a dice and tossing a coin.

Dependent events are events where the occurrence of one event affects the probability of another event.

Example:
Drawing two cards from a deck without replacement.

Understanding these concepts is important in probability theory, statistics, and real-life decision-making.

## Q7. What is the intuition of Bayes' Theorem in daily life?

Bayes' Theorem is a mathematical rule used to update probabilities when new information becomes available.

The main idea is:
Prior Knowledge + New Evidence = Updated Belief

For example, in medical diagnosis, a positive test result does not automatically mean a patient has a disease. Bayes' Theorem combines prior probability and test accuracy to determine the actual likelihood.

Bayesian reasoning is widely used in artificial intelligence, machine learning, medical diagnosis, and spam filtering.

## Q8. Explain Eigenvalue and Eigenvector in simple terms.

Eigenvalues and Eigenvectors are fundamental concepts in linear algebra.

When a matrix transforms a vector, most vectors change both direction and magnitude. However, some special vectors only change in magnitude and maintain the same direction.

These vectors are called Eigenvectors, and the scaling factor is called the Eigenvalue.

Mathematical Representation:

Av = λv

where:
A = Matrix
v = Eigenvector
λ = Eigenvalue

Eigenvalues and Eigenvectors are widely used in Principal Component Analysis (PCA), image compression, recommendation systems, and machine learning.


# Analyzing Income Distribution and Household Demographics

## Objective
This project analyzes household demographics, income levels, education, and distribution properties using Python.

The notebook includes:
- Types of data
- Descriptive statistics
- Central tendency
- Measures of dispersion
- Gaussian distribution
- Skewness and kurtosis
- Data visualization


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew, kurtosis, norm

# Load dataset
df = pd.read_csv('household_income_dataset.csv')

# Display first 5 rows
df.head()
```


# Part A: Theory & Definitions

## 1. Types of Data

### Numerical Data
Numerical data represents measurable quantities.

Examples from dataset:
- Age_of_Household_Head
- Household_Income
- Family_Size

### Categorical Data
Categorical data represents labels or groups.

Examples from dataset:
- Education_Level
- Owns_House
- Urban_Rural

---

## 2. Types of Statistics

### Descriptive Statistics
Descriptive statistics summarize and describe the dataset using:
- Mean
- Median
- Standard deviation
- Charts and graphs

### Inferential Statistics
Inferential statistics are used to make predictions or conclusions about a population based on sample data.

---

## 3. Mean, Median, and Mode

- Mean: Average value
- Median: Middle value
- Mode: Most frequent value

---

## 4. Range, Variance, and Standard Deviation

- Range = Maximum - Minimum
- Variance measures spread of data
- Standard deviation measures average distance from mean

---

## 5. Gaussian Distribution
A Gaussian (Normal) distribution is a bell-shaped distribution where most values lie near the mean.

---

## 6. Empirical Rule (68-95-99.7 Rule)

- 68% data lies within 1 standard deviation
- 95% data lies within 2 standard deviations
- 99.7% data lies within 3 standard deviations

---

## 7. Percentiles and Quartiles

- Quartiles divide data into 4 parts
- Percentiles divide data into 100 parts

---

## 8. Five Number Summary

- Minimum
- Q1
- Median
- Q3
- Maximum

---

## 9. Skewness
Skewness measures asymmetry of data.

---

## 10. Kurtosis
Kurtosis measures tail heaviness of distribution.


```python

# Identify Numerical and Categorical Columns

numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns
categorical_cols = df.select_dtypes(include=['object']).columns

print("Numerical Columns:")
print(numerical_cols)

print("\nCategorical Columns:")
print(categorical_cols)
```

```python

# Central Tendency

mean_income = df['Household_Income'].mean()
median_income = df['Household_Income'].median()
mode_income = df['Household_Income'].mode()[0]

mean_age = df['Age_of_Household_Head'].mean()
median_age = df['Age_of_Household_Head'].median()

print("Income Mean:", mean_income)
print("Income Median:", median_income)
print("Income Mode:", mode_income)

print("\nAge Mean:", mean_age)
print("Age Median:", median_age)
```

```python

# Measures of Dispersion

income_range = df['Household_Income'].max() - df['Household_Income'].min()
income_variance = df['Household_Income'].var()
income_std = df['Household_Income'].std()
iqr = df['Household_Income'].quantile(0.75) - df['Household_Income'].quantile(0.25)

print("Range:", income_range)
print("Variance:", income_variance)
print("Standard Deviation:", income_std)
print("Interquartile Range (IQR):", iqr)
```

```python

# Histogram of Household Income

plt.figure(figsize=(10,6))
sns.histplot(df['Household_Income'], kde=True)
plt.title('Histogram of Household Income')
plt.xlabel('Income')
plt.ylabel('Frequency')
plt.show()
```

```python

# Gaussian Distribution Curve

mu = df['Household_Income'].mean()
sigma = df['Household_Income'].std()

x = np.linspace(df['Household_Income'].min(),
                df['Household_Income'].max(), 100)

y = norm.pdf(x, mu, sigma)

plt.figure(figsize=(10,6))
plt.hist(df['Household_Income'], bins=20, density=True, alpha=0.6)
plt.plot(x, y, linewidth=2)
plt.title('Gaussian Distribution Curve')
plt.xlabel('Income')
plt.ylabel('Density')
plt.show()
```

```python

# Skewness and Kurtosis

income_skew = skew(df['Household_Income'])
income_kurtosis = kurtosis(df['Household_Income'])

print("Skewness:", income_skew)
print("Kurtosis:", income_kurtosis)
```

```python

# Boxplot: Income vs Education Level

plt.figure(figsize=(10,6))
sns.boxplot(x='Education_Level', y='Household_Income', data=df)
plt.title('Income Distribution Across Education Levels')
plt.show()
```

```python

# Family Size Distribution

plt.figure(figsize=(8,5))
sns.countplot(x='Family_Size', data=df)
plt.title('Family Size Distribution')
plt.show()
```

```python

# Age vs Income Scatter Plot

plt.figure(figsize=(10,6))
sns.scatterplot(x='Age_of_Household_Head',
                y='Household_Income',
                hue='Urban_Rural',
                data=df)

plt.title('Age vs Income')
plt.show()
```


# Conclusion

### Key Findings
- Graduate and Post-Graduate households generally have higher income.
- Income distribution is moderately spread.
- Histogram and Gaussian curve show near-normal distribution.
- Skewness and kurtosis help understand distribution shape.
- Urban households generally show higher income concentration.

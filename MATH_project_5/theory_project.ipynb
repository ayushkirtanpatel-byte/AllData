# 📘 Student Performance Analysis Using Linear Algebra

## 🎯 Introduction

This project analyzes the academic performance of 220 students using Linear Algebra concepts. The dataset contains marks in Math, Science, English, History, Computer Science, Physics, and Chemistry. The main objective is to apply vector operations, matrix operations, matrix decompositions, PCA, and LDA to understand patterns in student performance.

---

# 🔹 Part A: Vector Fundamentals

## 📌 Student Score Vectors

In Linear Algebra, a vector is an ordered collection of numbers.

Each student's marks can be represented as:

V = [Math, Science, English, History, Computer Science, Physics, Chemistry]

Example:

V = [78, 82, 75, 80, 88, 79, 84]

This vector represents the complete academic profile of a student.

### ✅ Norms of Vectors

### L1 Norm (Manhattan Norm)
The L1 norm is the sum of all vector values and represents the overall performance of a student.

### L2 Norm (Euclidean Norm)
The L2 norm measures the magnitude of a vector and helps compare students based on performance strength.

### ✅ Dot Product

The dot product measures similarity between two student vectors.

Higher Dot Product = More Similar Students

Applications:
- Student comparison
- Similarity analysis
- Recommendation systems

### ✅ Angle Between Vectors

The angle between vectors provides another measure of similarity.

Small Angle → Similar Performance

Large Angle → Different Performance

### ✅ Cross Product

Cross product was calculated using three subjects:

[Math, Science, English]

The resulting vector is perpendicular to both vectors and demonstrates geometric relationships in 3D space.

### ✅ Vector Projection

Vector projection determines how much one student's performance follows another student's performance pattern.

Applications:
- Student comparison
- Similarity detection
- Performance prediction

# 🔹 Part B: Matrix Fundamentals

## 📌 Student Subject Matrix

The dataset can be represented as a matrix.

Rows = Students (220)

Columns = Subjects (7)

Matrix Size = 220 × 7

Each element represents a student's score in a subject.

### ✅ Matrix Addition

Matrix addition combines corresponding elements of matrices.

Applications:
- Score aggregation
- Combining datasets

### ✅ Matrix Multiplication

Matrix multiplication is one of the most important operations in Linear Algebra.

Applications:
- Machine Learning
- Feature Transformations
- Data Analysis

### ✅ Matrix Transpose

Transpose swaps rows and columns.

Original Matrix:
220 × 7

Transpose Matrix:
7 × 220

Applications:
- Covariance Matrix
- PCA
- Statistical Analysis

### ✅ Matrix Inverse

The inverse matrix is used for solving systems of equations and advanced matrix calculations.

### ✅ Determinant

The determinant determines whether a matrix is invertible.

Determinant ≠ 0 → Invertible Matrix

Determinant = 0 → Singular Matrix

# 🔹 Part C: Linear Transformations & Geometry

## 📌 Line

A line represents one-dimensional space.

Example:
Variation in a single subject.

## 📌 Plane

A plane represents two-dimensional space.

Example:
Math vs Science performance.

## 📌 Hyperplane

A hyperplane extends the concept of a plane into higher dimensions.

Since the dataset contains seven subjects, each student exists in a 7-dimensional feature space.

Dimension Growth:

1D → Line

2D → Plane

3D → Space

4D+ → Hyperplane

# 🔹 Part D: Eigenvalues & Matrix Decomposition

## 📌 Covariance Matrix

The covariance matrix measures relationships among subjects.

Positive Covariance:
Both subjects increase together.

Negative Covariance:
One subject increases while another decreases.

### ✅ Eigenvalues

Eigenvalues represent the amount of variance captured by a specific direction.

Large Eigenvalue → More Important Information

Small Eigenvalue → Less Important Information

### ✅ Eigenvectors

Eigenvectors represent principal directions of variation within the dataset.

Applications:
- PCA
- Machine Learning
- Pattern Recognition

### ✅ LU Decomposition

LU decomposition factorizes a matrix:

A = L × U

Where:

L = Lower Triangular Matrix

U = Upper Triangular Matrix

Applications:
- Solving Linear Equations
- Matrix Inversion
- Numerical Computation

### ✅ Singular Value Decomposition (SVD)

SVD factorizes a matrix:

A = UΣVᵀ

Applications:
- Data Compression
- Noise Reduction
- Recommendation Systems
- Machine Learning

---

# 🔹 Part E: Dimensionality Reduction

## 📌 Principal Component Analysis (PCA)

PCA reduces seven-dimensional data into two principal components while preserving maximum variance.

Benefits:

✔ Better Visualization

✔ Noise Reduction

✔ Faster Computation

✔ Feature Extraction

### ✅ Linear Discriminant Analysis (LDA)

LDA separates students into:

- Above Average
- Below Average

Benefits:

✔ Better Classification

✔ Reduced Dimensions

✔ Improved Interpretation

This project demonstrates the practical application of Linear Algebra concepts on student performance data.
